## Exception Handling

Errors: these occur due to mistake in the code that prevent it from running. these can be syntax error or logical error
There are many types of error which can not be handled
1. syntax error
2. indentation error

Exceptions: these are the unexpected events or error that occurs during the execution of the progeam, whicg disrupts the normal flow of the program

```python
a = int(input("tell your number :"))

print(10/a)
```

```python
## if thee user gives the 0 as the input then it gives the error of division by zero.
## so to handle this error we use try and except block

try:
    a = int(input("tell your number :"))
    print(10/a)
except ZeroDivisionError:
    print("You can't divide by zero")
```

```python
# try: wrap the block of code that might cause the exception
# except: handle the exception if it occurs
# else: run code if no exception occurs
# finally: run code no matter what, whether thats an exception or not
# raise: manually throw an exception
```

```python
## we will not be knowing what error might come in the code so we can use the except block without specifying any error name. It will handle all the errors.
try:
    a = int(input("tell your number :"))
    print(10/a)
except Exception as err:
    print("sorry, This is an error:", err)
```

```python
try:
    a = int(input("tell your number :"))
    print(10/a)
except Exception as err:
    print("sorry, This is an error:", err)
# if except block dosent run then else block will run
else:
    print("great, there is no exception")
```

```python
try:
    a = int(input("tell your number :"))
    print(10/a)
except Exception as err:
    print("sorry, This is an error:", err)
# if except block dosent run then else block will run
else:
    print("great, there is no exception")
# finnaly is block which will run no matter what
finally:
    print("I am finally block, I run no matter what loll")
```

```python
# raise keyword is used to manually raise an exception of our choice
age = int(input("tell your age :"))
if age < 18:
    raise ValueError("Age is below 18, You are not allowed")
```
