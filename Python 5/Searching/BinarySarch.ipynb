## Binary Search

->the elements should be sorted before doing the binary search

### Steps:

 1. consider the full array or list. where start = 0 and end = n-1 

 2. see the middle element 
        middle = (start + end) // 2.

 3. compare the middle element with target 

 4. if target == middle element return index of middle element 

 5. if target is greater than midlle element. Set the start = middle element+1 and end = n-1 and repeat the step 2, 3 and 4. proceed this step untill the target == middle element or target is lesser than middle element

 6. if target is lesser than midlle element. Set the start = 0 and end = middle element - 1 and repeat the step 2, 3 and 4. proceed this step until the target == middle element or target is greater than middle element

```python
## Basic code implementation
def binary_search(list, target):
    start = 0
    end = len(list)
    while start <= end :
        middle = (start + end) // 2
        if list[middle] == target :
            return middle
        elif list[middle] < target :
            start = middle +1
        elif list[middle] > target : 
            end = middle -1
    return -1




list = [1,2,4,5,7,9,12,14,16,120,235,457]
target = int(input("Enter the target number/element"))
result = binary_search(list, target)

if result != -1:
    print(f"Element found at index: {result}")
else:    print("Element not found in the list.")
```

```python
## code implementation using Recursion
def binary_search_recursion(list1, target1, start, end):
    if start > end:
        return -1
    middle = (start + end) // 2
    if list1[middle] == target1:
        return middle
    elif list1[middle] < target1:
        return binary_search_recursion(list1, target1, middle + 1, end)
    else:
        return binary_search_recursion(list1, target1, start, middle - 1)
    


list1 = [1,2,4,5,7,9,12,14,16,120,235,457]
start = 0
end = len(list1)
target1 = int(input("Enter the target number/element"))
result = binary_search_recursion(list1, target1, start, end)

if result != -1:
    print(f"Element found at index: {result}")
else:    print("Element not found in the list.")
```
