### Linear Search:

->Here we will be given an array and target

->we have to return the target index if present. else, return -1.

->we need to run the loop and compaire the target and array element

```python
def linear_search(my_list, target):
    for i in range(len(my_list)):
        if my_list[i] == target:
            return i
    return -1

my_list =[10,342,54,23,76,23]

target = int(input("Enter the target number/element"))

result = linear_search(my_list, target)

if result != -1:
    print(f"Element found at index: {result}")  
else:    print("Element not found in the list.")
```
