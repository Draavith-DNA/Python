## Bubble sorting

 ### Steps

 1. compare 0th and 1st elements in array
 2. if arr[0] > arr[1], Then swap. Else Goto step 3
 3. compare 1st and 2nd element in array. then swap if needed else go to next comparision
4. repeate these comare and swap unless the array is sorted compleatly

-> if we have n number of elements we do n-1 number of comparision in first itteration

-> It is guarantee that in 1st step we will push/sort the largest element to the end

### vilualization:
https://www.hackerearth.com/practice/algorithms/sorting/bubble-sort/visualize/

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1): # Last i elements are already in place
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j] # Swap if the element found is greater than the next element
    return arr



unsorted_list = [23,64,23,564,76,2,45,4]
sorted_list = bubble_sort(unsorted_list)
print("Sorted Elements:", sorted_list)
```
